## 2. What a token is
A **tokenizer** is a fixed dictionary that splits text into pieces the model can read. The pieces are **tokens** — roughly ¾ of an English word each. The dictionary was built mostly from English and code, so other scripts cost more.

Our default model (`openai/gpt-oss-20b`) uses the `o200k_harmony` encoding, which `tiktoken` ships — so for this model the count is **exact**. For other models it is an estimate; the model's own `usage` field is the truth after you send (section 3).

```python
TEXTS = {
    "english": "Can I get a personal loan of 8 lakh rupees for five years?",
    "telugu":  "నాకు ఐదు సంవత్సరాలకు 8 లక్షల రూపాయల వ్యక్తిగత రుణం లభిస్తుందా?",
    "python":  "def emi(p, r, n):\n    r = r/12/100\n    return p*r*(1+r)**n/((1+r)**n-1)",
    "json":    '{"pan": "ABCDE1234F", "amount": 800000, "tenure_months": 60}',
}
print(f"{'text':8} {'chars':>6} {'tokens':>7} {'chars/token':>12}")
for name, t in TEXTS.items():
    n = count_tokens(t)
    print(f"{name:8} {len(t):6d} {n:7d} {len(t)/n:12.1f}")
```

```python
# See the pieces. Where does the tokenizer cut?
for name in ("english", "telugu"):
    print(name, "→", show_tokens(TEXTS[name]), "\n")
```

## 3. Check against reality — the `usage` field
Send the same texts to a real model and read `usage.prompt_tokens`. The model adds a few *template tokens* around your message (role markers), so expect the real number to be slightly higher than the raw count. If Ollama is running, send them there too — a **different model, a different tokenizer, a different count**.

```python
def usage(client, model, text):
    return usage_for(client, model, text)[0]


overhead_groq = usage(groq, MODEL, "hi") - count_tokens("hi")
print(f"template wrapper on Groq/{MODEL}: {overhead_groq} tokens you never typed\n")
print(
    f"{'text':8} {'tiktoken':>9} {'groq usage':>11} {'minus wrapper':>14} {'ollama usage':>13}"
)
for name, t in TEXTS.items():
    est = count_tokens(t)
    g = usage(groq, MODEL, t)
    try:
        o = str(usage(ollama, "llama3.2:3b", t))
    except (ConnectionError, TimeoutError, RuntimeError):
        o = "(no ollama)"
    print(f"{name:8} {est:9d} {g:11d} {g - overhead_groq:14d} {o:>13}")
```

## 5. The context window as a budget
A model sees a fixed number of tokens at once — its **context window**. Everything must fit: your system prompt, the whole conversation so far, any documents you paste in, any tool results. Here we simulate a growing conversation and watch the count climb. Where the line crosses the limit, something has to be dropped — that is Session 19's problem (memory windows) and Session 38's (context engineering).

```python
import matplotlib.pyplot as plt
system = "You are a concise assistant for a retail bank."
history = [{"role": "system", "content": system}]
totals = []
for turn in range(1, 21):
    history.append({"role": "user", "content": f"Turn {turn}: {TEXTS['english']}"})
    history.append({"role": "assistant", "content": "Certainly. " + "Here is a detailed answer about your loan. " * 6})
    totals.append(sum(count_tokens(m["content"]) for m in history))
LIMIT = 8192   # a small window, for the picture; real models are far larger — but so are real conversations with documents
plt.plot(range(1, 21), totals, marker="o"); plt.axhline(LIMIT, color="red", linestyle="--", label=f"limit {LIMIT}")
plt.xlabel("turn"); plt.ylabel("tokens in context"); plt.title("The conversation grows every turn"); plt.legend(); plt.show()
print("tokens at turn 20:", totals[-1])
```

## 7. Reflection
Answer the five questions above here, one or two lines each.

1. Even though the Telugu sentence has the same meaning as the English sentence, it uses more tokens because the tokenizer handles English more efficiently than Telugu. In the given example, the English sentence has **58 characters and 16 tokens**, which is about **3.6 characters per token**. The Telugu sentence has **62 characters but 25 tokens**, only about **2.5 characters per token**.

    This happens because the model's tokenizer was trained mostly on English and code, so common English words can often be represented using larger token pieces. Telugu text is more often split into smaller sub-word or character-level pieces. As a result, the same information needs more token IDs.

     For a bank with many Telugu-speaking customers, this means higher token usage and potentially higher LLM costs. In this example, Telugu uses about **56% more tokens** than English (25 vs. 16). The larger token count also means Telugu conversations consume the model's context window faster, leaving less room for long conversations or documents. At large scale, this difference can become significant for a bank serving many Telugu customers.


2. The Telugu row shows this most clearly. The exact same Telugu sentence was counted as **25 tokens using the Groq tokenizer**, but when the same text was processed by the **Ollama Llama model, it used 132 tokens**. The text itself did not change, only the model and its tokenizer changed. This shows that tokens are not a fixed property of the text; they depend on how a particular model's tokenizer breaks the text into pieces. Different models can therefore use very different numbers of tokens for the same sentence.


3. Since output tokens cost more than input tokens, we should ask the model to give **clear and concise answers instead of unnecessarily long responses**. We should provide specific instructions about the required length and avoid asking for extra details that are not needed. This helps reduce the number of output tokens and therefore keeps the cost lower, especially when the model is used many times.


4. At turn 21, the model's context window contains the **system prompt, the conversation so far, documents or other information provided, and tool results**. As the conversation grows, the total number of tokens can reach the context-window limit. When that happens, some earlier information has to be dropped to make space for the newer content. **The model/system handling the context window decides what has to be dropped**, not the user or the text itself.


5. The banking words — **loan, bank, interest, EMI, and credit score** — form a clear cluster because they are related to the same financial context. Similarly, **hospital, doctor, and prescription** are grouped together because they are connected to healthcare. **Cricket and monsoon** are positioned separately because their meanings are not closely related to either banking or healthcare.

    An embedding model cannot **generate text, answer questions, or reason about information**. Its job is to convert text into numerical vectors (embeddings), where semantically similar words tend to have nearby coordinates. It represents meaning as coordinates, but it does not itself produce an answer.
